Este dataset tiene información de 6,607 estudiantes con datos sobre sus hábitos de estudio, su situación familiar y su entorno escolar, y busca ver qué tanto influyen esos factores en su calificación final de examen.

Cada fila es un estudiante distinto, y cada columna es una característica de ese estudiante (horas que estudia, asistencia, si tiene acceso a internet, etc.) junto con el puntaje que sacó en su examen (Exam_Score).

La variable objetivo sería Exam_Score, porque es un número que representa el resultado final y es lo que se buscaría predecir.

Como Exam_Score es un número continuo, este sería un problema de regresión.

Un modelo entrenado con estos datos podría servir para detectar desde antes qué estudiantes van a tener un desempeño bajo, y así poder ayudarlos con tutorías u otro tipo de apoyo antes del examen.

Los tipos de datos están bien: 6 columnas son números enteros y 13 son texto/categorías. No hay registros duplicados. Sí hay valores faltantes en 3 columnas: Teacher_Quality (78), Parental_Education_Level (90) y Distance_from_Home (67), aunque en los tres casos es menos del 1.5% de los datos. En las estadísticas también se ve que Exam_Score llega hasta 101, lo cual es raro porque un examen normalmente no debería pasar de 100.

```python
# esto junta las columnas que son de texto
columnas_texto = df.select_dtypes(include='object').columns

# esto revisa si alguna columna de texto en realidad trae puros números
numeric_like_text = [c for c in columnas_texto
                      if pd.to_numeric(df[c], errors='coerce').notna().mean() > 0.9]
print("Columnas de texto que en realidad parecen numéricas:", numeric_like_text)

# esto cuenta si hay espacios de más al inicio o final del texto en cada columna
espacios = {}
for c in columnas_texto:
    sin_nulos = df[c].dropna().astype(str)
    espacios[c] = int((sin_nulos != sin_nulos.str.strip()).sum())
print("Celdas con espacios de más por columna:", {k: v for k, v in espacios.items() if v > 0} or "Ninguna")

# esto imprime todas las categorías distintas que tiene cada columna de texto
for c in columnas_texto:
    print(c, ":", sorted(df[c].dropna().unique()))

# esto busca columnas donde todos los valores son iguales (no aportan nada)
constantes = [c for c in df.columns if df[c].nunique() == 1]
print("\nVariables constantes:", constantes or "Ninguna")

# esto busca si alguna columna es como un ID único por cada fila
identificadores = [c for c in df.columns if df[c].nunique() == len(df)]
print("Posibles identificadores:", identificadores or "Ninguno")

# esto cuenta cuántas filas tienen Exam_Score arriba de 100
print("\nRegistros con Exam_Score > 100:", (df['Exam_Score'] > 100).sum())

# esto muestra esas filas
df[df['Exam_Score'] > 100]
```

```python
# esto junta solo las columnas numéricas
columnas_numericas = df.select_dtypes(include=np.number).columns

# esto revisa columna por columna si hay valores muy alejados del resto
for c in columnas_numericas:
    # esto calcula en qué valor caen el 25% y el 75% de los datos
    q1, q3 = df[c].quantile([0.25, 0.75])

    # esto saca el rango entre esos dos puntos
    iqr = q3 - q1

    # esto calcula el límite de abajo y de arriba para considerar algo "raro"
    lim_inf = q1 - 1.5 * iqr
    lim_sup = q3 + 1.5 * iqr

    # esto cuenta cuántos valores se salen de esos límites
    outliers = ((df[c] < lim_inf) | (df[c] > lim_sup)).sum()

    print(f"{c}: {outliers} posibles valores raros (límites: {lim_inf:.1f} a {lim_sup:.1f})")
```

```python
# esto hace una copia del dataset original para no perderlo
df_clean = df.copy()
```

```python
# esto quita espacios de más en los nombres de las columnas, por si acaso
df_clean.columns = df_clean.columns.str.strip()
print("Nombres de columnas revisados, no hubo cambios necesarios.")
```

```python
# esto guarda cuántas filas hay antes de quitar duplicados
antes = len(df_clean)

# esto elimina filas que estén completamente repetidas
df_clean = df_clean.drop_duplicates()

print(f"Filas antes: {antes} | filas después: {len(df_clean)}")
```

```python
# esto junta las columnas de texto
columnas_texto = df_clean.select_dtypes(include='object').columns

# esto convierte esas columnas a tipo "categoría", que es más correcto para este tipo de datos
for c in columnas_texto:
    df_clean[c] = df_clean[c].astype('category')

print(df_clean.dtypes)
```

```python
# esto quita espacios de más en el texto de cada celda, por si acaso
for c in columnas_texto:
    df_clean[c] = df_clean[c].astype(str).str.strip()
    # esto arregla que al convertir a texto, los nulos se vuelven la palabra "nan"
    df_clean.loc[df_clean[c] == 'nan', c] = np.nan
```

```python
# esto junta las 3 columnas que tenían valores faltantes
columnas_con_faltantes = ['Teacher_Quality', 'Parental_Education_Level', 'Distance_from_Home']

for c in columnas_con_faltantes:
    # esto saca cuál es el valor que más se repite en esa columna
    moda = df_clean[c].mode()[0]

    # esto cuenta cuántos valores faltaban antes de rellenar
    n_faltantes = df_clean[c].isnull().sum()

    # esto rellena los valores vacíos con el valor que más se repite
    df_clean[c] = df_clean[c].fillna(moda)

    print(f"{c}: se rellenaron {n_faltantes} valores con '{moda}'")

print("\nValores faltantes que quedan:", df_clean.isnull().sum().sum())
```

```python
# esto guarda cuántas filas hay antes de quitar el valor imposible
antes = len(df_clean)

# esto se queda solo con las filas donde Exam_Score es 100 o menos
df_clean = df_clean[df_clean['Exam_Score'] <= 100].reset_index(drop=True)

print(f"Filas antes: {antes} | filas después: {len(df_clean)}")
```

Se revisaron los nombres de columnas (no hizo falta cambiar nada), se confirmó que no había duplicados, se pasaron las columnas de texto a tipo categoría, se rellenaron los valores faltantes de Teacher_Quality, Parental_Education_Level y Distance_from_Home con el valor más común de cada una, y se quitó la fila que tenía un Exam_Score de 101 porque un examen no puede pasar de 100. No se quitó ninguna otra fila ni columna porque el resto de los datos raros que se encontraron sí son posibles.

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

# esto junta las columnas que sí tienen un orden claro
ordinal_cols = ['Parental_Involvement', 'Access_to_Resources', 'Motivation_Level',
                 'Family_Income', 'Teacher_Quality', 'Distance_from_Home',
                 'Parental_Education_Level']

# esto define, para cada columna de arriba, cuál es el orden correcto de sus categorías
ordinal_categories = [
    ['Low', 'Medium', 'High'],
    ['Low', 'Medium', 'High'],
    ['Low', 'Medium', 'High'],
    ['Low', 'Medium', 'High'],
    ['Low', 'Medium', 'High'],
    ['Near', 'Moderate', 'Far'],
    ['High School', 'College', 'Postgraduate'],
]

# esto junta las columnas que no tienen ningún orden entre sus categorías
nominal_cols = ['Extracurricular_Activities', 'Internet_Access', 'School_Type',
                 'Peer_Influence', 'Learning_Disabilities', 'Gender']

print("Ordinales:", ordinal_cols)
print("Nominales:", nominal_cols)
```

## 7. Codificación de variables categóricas

```python
# esto crea el codificador ordinal con el orden que definimos arriba
ordinal_encoder = OrdinalEncoder(categories=ordinal_categories)

# esto aplica el codificador y convierte el resultado en tabla
ejemplo_ordinal = pd.DataFrame(
    ordinal_encoder.fit_transform(df_clean[ordinal_cols]),
    columns=ordinal_cols
)

# esto junta las columnas originales con su versión ya codificada, para comparar
pd.concat([df_clean[ordinal_cols].reset_index(drop=True).head(),
           ejemplo_ordinal.head().add_suffix('_encoded')], axis=1)
```

```python
# esto crea el codificador One-Hot (quita la primera categoría de cada variable para no repetir información)
onehot_encoder = OneHotEncoder(drop='first', sparse_output=False)

# esto aplica el codificador y convierte el resultado en tabla
ejemplo_onehot = pd.DataFrame(
    onehot_encoder.fit_transform(df_clean[nominal_cols]),
    columns=onehot_encoder.get_feature_names_out(nominal_cols)
)

ejemplo_onehot.head()
```

La diferencia es que en la codificación ordinal, el número sí representa un orden real: Low=0, Medium=1, High=2, y el modelo puede entender que High es más que Low. En cambio, si a School_Type se le pusiera un número así, el modelo pensaría que "Private" es más grande que "Public", lo cual no tiene sentido porque no hay ningún orden entre esas dos categorías. Por eso a las nominales se les hace One-Hot Encoding, que crea una columna de sí/no por cada categoría en lugar de un solo número.

## 8. Escalamiento de variables numéricas

Se aplica StandardScaler a Hours_Studied y Attendance, para que ambas queden con media 0 y desviación estándar 1.

```python
from sklearn.preprocessing import StandardScaler

# esto crea el escalador
scaler_demo = StandardScaler()

# esto junta las columnas que se van a escalar
cols_a_escalar = ['Hours_Studied', 'Attendance']

# esto guarda las estadísticas de esas columnas antes de escalar
antes_escalado = df_clean[cols_a_escalar].describe().loc[['mean', 'std', 'min', 'max']]

# esto aplica el escalador y saca las estadísticas después de escalar
despues_escalado = pd.DataFrame(
    scaler_demo.fit_transform(df_clean[cols_a_escalar]),
    columns=cols_a_escalar
).describe().loc[['mean', 'std', 'min', 'max']]

print("Antes del escalamiento:")
print(antes_escalado)
print("\nDespués del escalamiento:")
print(despues_escalado)
```

Después de escalar, las dos columnas quedan con media casi 0 y desviación estándar 1, pero conservan la forma de su distribución y el orden entre los estudiantes; solo cambia la escala en la que se miden. Esto importa porque algoritmos como KNN, SVM o redes neuronales calculan distancias basadas en el tamaño de los números, y si Attendance (60-100) se deja sin escalar junto con Hours_Studied (1-44), la columna con números más grandes dominaría el cálculo sin que eso signifique que en realidad es más importante.

## 9. Organización reproducible con ColumnTransformer


```python
from sklearn.compose import ColumnTransformer

# esto junta las columnas numéricas que se van a escalar
numeric_cols = ['Hours_Studied', 'Attendance', 'Sleep_Hours',
                 'Previous_Scores', 'Tutoring_Sessions', 'Physical_Activity']

# esto arma un solo bloque que aplica escalamiento a las numéricas,
# codificación ordinal a las que tienen orden, y One-Hot a las que no
preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), numeric_cols),
    ('ord', OrdinalEncoder(categories=ordinal_categories), ordinal_cols),
    ('nom', OneHotEncoder(drop='first', sparse_output=False), nominal_cols),
], remainder='drop')

# esto separa las columnas que se van a transformar (X) de la columna objetivo (y)
X = df_clean.drop(columns=['Exam_Score'])
y = df_clean['Exam_Score'].reset_index(drop=True)

# esto aplica todas las transformaciones de una sola vez
X_transformed = preprocessor.fit_transform(X)

# esto saca los nombres de las columnas ya transformadas
feature_names = preprocessor.get_feature_names_out()

# esto arma la tabla final ya transformada, y le regresa Exam_Score
df_prepared = pd.DataFrame(X_transformed, columns=feature_names)
df_prepared['Exam_Score'] = y

print("Forma del dataset transformado:", df_prepared.shape)
df_prepared.head()
```

## 12. Conclusión

El diagnóstico inicial del dataset encontró tres problemas principales: valores faltantes en Teacher_Quality, Parental_Education_Level y Distance_from_Home (todos por debajo del 1.5% de los datos), un valor imposible en Exam_Score (101 puntos, cuando el máximo es 100), y algunos valores atípicos en variables como Exam_Score y Hours_Studied que se dejaron documentados pero no se quitaron porque son valores reales y posibles. No se encontraron duplicados, columnas numéricas guardadas como texto, categorías mal escritas, variables constantes ni columnas de identificador.

Las transformaciones más importantes fueron: rellenar los valores faltantes con la moda, quitar el registro con el puntaje imposible, codificar de forma ordinal las 7 variables que sí tienen un orden claro (como Parental_Involvement), codificar con One-Hot las 6 variables que no tienen orden (como School_Type o Gender), y escalar 6 variables numéricas con StandardScaler. La codificación permitió pasar todas las variables de texto a números que un modelo pueda usar, respetando el orden real en las que sí lo tienen. El escalamiento evitó que las variables con rangos más grandes (como Attendance) dominen sobre las que tienen rangos más chicos (como Sleep_Hours).

Como limitación, el dataset todavía tiene algunos valores atípicos que no se quitaron, y las transformaciones de este notebook se ajustaron usando todos los datos, solo para fines de esta actividad. El siguiente paso antes de entrenar un modelo sería dividir los datos en entrenamiento y prueba, y ajustar el ColumnTransformer solamente con los datos de entrenamiento, para no meter información del conjunto de prueba y así evitar la fuga de información.
